```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = '/content/employee_information_100.csv'

df = pd.read_csv(file_path)

print("=" * 60)
print("EMPLOYEE DATASET")
print("=" * 60)
print(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

df.columns = df.columns.str.strip()

print("\nCleaned Column Names:")
print(df.columns.tolist())


def find_column(possible_names):
    column_map = {
        col.lower().replace(" ", "").replace("_", ""): col
        for col in df.columns
    }
    for name in possible_names:
        key = name.lower().replace(" ", "").replace("_", "")
        if key in column_map:
            return column_map[key]
    return None


department_col = find_column(['Department', 'Dept', 'DepartmentName'])
salary_col = find_column(['Salary', 'AnnualSalary', 'Income', 'Pay'])
gender_col = find_column(['Gender', 'Sex'])
experience_col = find_column([
    'Experience',
    'YearsExperience',
    'YearsOfExperience',
    'ExperienceYears',
    'WorkExperience'
])
age_col = find_column(['Age', 'EmployeeAge'])

print("\n" + "=" * 60)
print("DETECTED COLUMNS")
print("=" * 60)
print("Department :", department_col)
print("Salary     :", salary_col)
print("Gender     :", gender_col)
print("Experience :", experience_col)
print("Age        :", age_col)

required_columns = {
    'Department': department_col,
    'Salary': salary_col,
    'Gender': gender_col,
    'Experience': experience_col,
    'Age': age_col
}

missing = [
    name for name, column in required_columns.items()
    if column is None
]

if missing:
    print("\n" + "=" * 60)
    print("ERROR")
    print("=" * 60)
    print("The following columns could not be detected:")
    print(missing)
    print("\nYour actual columns are:")
    print(df.columns.tolist())
    print("\nPlease check the column names in your dataset.")

else:
    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.1")
    print("=" * 60)

    average_salary = df.groupby(department_col)[salary_col].mean()

    print("Average Salary of Each Department:")
    print(average_salary)

    plt.figure(figsize=(8, 5))
    average_salary.plot(kind='bar')
    plt.title('Average Salary of Each Department')
    plt.xlabel('Department')
    plt.ylabel('Average Salary')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.2")
    print("=" * 60)

    department_count = df[department_col].value_counts()

    print("Number of Employees in Each Department:")
    print(department_count)

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.3")
    print("=" * 60)

    gender_count = df[gender_col].value_counts()
    gender_percentage = (gender_count / len(df)) * 100

    print("Gender Count:")
    print(gender_count)
    print("\nGender Percentage:")
    print(gender_percentage)

    plt.figure(figsize=(7, 7))
    plt.pie(
        gender_percentage,
        labels=gender_percentage.index,
        autopct='%1.1f%%',
        startangle=90
    )
    plt.title('Percentage of Male and Female Employees')
    plt.show()

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.4")
    print("=" * 60)

    plt.figure(figsize=(8, 5))
    plt.hist(df[salary_col], bins=10, edgecolor='black')
    plt.title('Salary Distribution')
    plt.xlabel('Salary')
    plt.ylabel('Number of Employees')
    plt.tight_layout()
    plt.show()

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.5")
    print("=" * 60)

    plt.figure(figsize=(8, 5))
    plt.scatter(df[experience_col], df[salary_col])
    plt.title('Relationship Between Experience and Salary')
    plt.xlabel('Years of Experience')
    plt.ylabel('Salary')
    plt.tight_layout()
    plt.show()

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.6")
    print("=" * 60)

    top_10 = df.nlargest(10, salary_col)

    print("Top 10 Employees with Highest Salary:")
    print(top_10)

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.7")
    print("=" * 60)

    highest_salary = df.groupby(department_col)[salary_col].max()

    print("Highest Salary in Every Department:")
    print(highest_salary)

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.8")
    print("=" * 60)

    overall_average_salary = df[salary_col].mean()

    print("Overall Average Salary:")
    print(overall_average_salary)

    employees_above_average = df[df[salary_col] > overall_average_salary]

    print("\nEmployees with Salary Greater Than Overall Average:")
    print(employees_above_average)

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.9")
    print("=" * 60)

    average_experience = df.groupby(department_col)[experience_col].mean()

    print("Average Years of Experience in Each Department:")
    print(average_experience)

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.10")
    print("=" * 60)

    plt.figure(figsize=(8, 5))
    plt.hist(df[age_col], bins=10, edgecolor='black')
    plt.title('Distribution of Employee Ages')
    plt.xlabel('Age')
    plt.ylabel('Number of Employees')
    plt.tight_layout()
    plt.show()

    print("\n\n" + "=" * 60)
    print("EXPERIMENT 1.1 TO 1.10 COMPLETED")
    print("=" * 60)
```
