## 2.1 Text cleaning

```python
from collections import Counter

PUNCTUATION = r"""!"#$%&'()*+,-./:;<=>?@[\]^_`{|}~"""

text = open('2.1_text_data.txt', encoding='utf-8').read()
print(text)
```

Count what we are about to strip out. Digits and whitespace are counted in groups,
so 12345 is one number and three spaces in a row is one gap.

```python
uppercase = [c for c in text if c.isupper()]
punctuation = [c for c in text if c in PUNCTUATION]

numbers = []
spaces = []
digits = ''
gap = ''

for c in text:
    if c.isdigit():
        digits += c
    elif digits:
        numbers.append(digits)
        digits = ''

    if c.isspace():
        gap += c
    else:
        if len(gap) > 1:
            spaces.append(gap)
        gap = ''

if digits:
    numbers.append(digits)
if len(gap) > 1:
    spaces.append(gap)

print('Uppercase letters:', len(uppercase))
print('Punctuation marks:', len(punctuation))
print('Punctuation types:', Counter(punctuation))
print('Numbers:', len(numbers), '->', numbers)
print('Places with extra whitespace:', len(spaces))
print('Extra whitespace characters:', sum(len(s) - 1 for s in spaces))
```

```python
clean = text.lower()
clean = ''.join(c for c in clean if c not in PUNCTUATION)
clean = ''.join(c for c in clean if not c.isdigit())
clean = ' '.join(clean.split())  # split() drops every run of whitespace

print(clean)
```

## 2.2 Tokenization

Same text, three ways.

```python
text = open('2.2_tokenization_data.txt', encoding='utf-8').read()
print(text)
```

### Using NLTK

```python
from nltk.tokenize import sent_tokenize, word_tokenize

sentences = sent_tokenize(text)
words = word_tokenize(text)

print('Sentences:', len(sentences))
for i, s in enumerate(sentences, 1):
    print(i, s)

print()
print('Words:', len(words))
print(words)
```

### Using spaCy

```python
import spacy

nlp = spacy.load('en_core_web_sm')
doc = nlp(text)

sentences = [s.text.strip() for s in doc.sents]
words = [t.text for t in doc if not t.is_space]

print('Sentences:', len(sentences))
for i, s in enumerate(sentences, 1):
    print(i, s)

print()
print('Words:', len(words))
print(words)
```

### Without NLTK or spaCy

```python
# a sentence ends at . ! or ?
sentences = []
current = ''

for c in text:
    current += c
    if c in '.!?':
        sentences.append(current.strip())
        current = ''

if current.strip():
    sentences.append(current.strip())

print('Sentences:', len(sentences))
for i, s in enumerate(sentences, 1):
    print(i, s)
```

```python
# a word runs until a character that is not a letter, digit or apostrophe
words = []
current = ''

for c in text:
    if c.isalnum() or c == "'":
        current += c
    elif current:
        words.append(current)
        current = ''

if current:
    words.append(current)

print('Words:', len(words))
print(words)
```

## 2.3 Stop word removal

```python
from nltk.corpus import stopwords

text = open('2.3_clean_data.txt', encoding='utf-8').read()
stop_words = set(stopwords.words('english'))

# drop punctuation tokens so only real words are compared
tokens = [t for t in word_tokenize(text) if t.isalpha()]
filtered = [t for t in tokens if t.lower() not in stop_words]
removed = [t for t in tokens if t.lower() in stop_words]

print('Original tokens:', len(tokens))
print(tokens)
```

```python
print('Filtered tokens:', len(filtered))
print(filtered)

print()
print('Stop words removed:', len(removed))
print(removed)

print()
print('Stop word counts:', Counter(w.lower() for w in removed))
```
